```python
import requests
import matplotlib.pyplot as plt
import geopandas as gpd
import cartopy.crs as ccrs
import cartopy.feature as cfeature
```

```python
gdf = gpd.read_file('http://localhost:8000/api/v1/geometries/WARNINGS/by_type?type_name=freeze&key=20260421_0800')
print(gdf)
gdf = gpd.read_file('http://localhost:8000/api/v1/geometries/WARNINGS/by_type?type_name=flash_flood&key=20260420_2200')
print(gdf)
```

```python
# 1. Setup projection
proj = ccrs.PlateCarree()

# 2. Create the plot with Cartopy axes
fig, ax = plt.subplots(figsize=(10, 6), subplot_kw={'projection': proj})

# 3. Add Cartopy basemap features (borders, coastlines)
ax.add_feature(cfeature.BORDERS, linestyle=':')
ax.add_feature(cfeature.COASTLINE)
ax.add_feature(cfeature.STATES)

gdf.plot(ax=ax)
```
