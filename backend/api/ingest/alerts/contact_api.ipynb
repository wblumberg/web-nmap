```python
import requests
import json
import geopandas as gpd
import cap_parser
```

```python
url = 'https://api.weather.gov/alerts/active?status=actual'
response = requests.get(url)
data = response.json()
```

```python
from shapely.geometry import shape
from datetime import datetime, timezone

for feature in data['features']:
    properties = feature['properties']
    #if 'Severe Thunderstorm Watch' not in properties['event']:
    if 'Watch' not in properties['event']:

        continue    
    print(f"Alert: {properties['event']}")
    product_polygon = feature['geometry']
    same = properties['geocode']['SAME']
    print(properties)
    print("Product Polygon:", product_polygon)
    print("Message Type:", properties['messageType'])
    print("References: ", properties['references'])
    print('VTEC:', properties['parameters']['VTEC'])
    print("Parsed VTEC:", cap_parser.parse_pvtec_string(properties['parameters']['VTEC'][0]))
    counties['FIPS'] = counties['FIPS'].astype(str).str.zfill(5)
    same_codes = [code[-5:] for code in same]  # SAME is 6-digit; county FIPS is last 5
    same = counties['FIPS'].where(counties['FIPS'].isin(same_codes), other='__NO_MATCH__')
    county_outlines = counties[counties['FIPS'] == same][['FIPS', 'geometry']]['geometry'].union_all()
    if county_outlines.is_empty:
        county_outlines = "No matching counties found"
    else:
        import matplotlib.pyplot as plt

        alert_geom = shape(product_polygon) if product_polygon else None
        matched_counties = counties[counties['FIPS'].isin(same_codes)]
        state_borders = counties.dissolve(by='STATE').boundary

        fig, ax = plt.subplots(figsize=(10, 10))

        counties.boundary.plot(ax=ax, color='lightgray', linewidth=0.3)
        state_borders.plot(ax=ax, color='black', linewidth=1.0)
        matched_counties.boundary.plot(ax=ax, color='blue', linewidth=1.2)
        gpd.GeoSeries([county_outlines]).boundary.plot(ax=ax, color='green', linewidth=2.0)

        if alert_geom is not None and not alert_geom.is_empty:
            gpd.GeoSeries([alert_geom]).boundary.plot(ax=ax, color='red', linewidth=2.0)

        # Zoom to CONUS extent
        ax.set_xlim(-125, -66.5)
        ax.set_ylim(24, 49.5)

        ax.set_title(f"{properties['event']} - Polygon / Counties / State Borders")
        ax.set_axis_off()
        plt.show()
    #print("County Outlines:", county_outlines)
    def _to_utc(ts):
        if not ts:
            return None
        return datetime.fromisoformat(ts.replace('Z', '+00:00')).astimezone(timezone.utc).isoformat().replace('+00:00', 'Z')

    print(f"Onset (UTC): {_to_utc(properties.get('onset'))}")
    print(f"Effective (UTC): {_to_utc(properties.get('effective'))}")
    print(f"Expires (UTC): {_to_utc(properties.get('expires'))}")
    print("-" * 40)
```
